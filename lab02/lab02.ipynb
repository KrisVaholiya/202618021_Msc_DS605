## Task 2 - Vectorized Arithmetic and Linear Algebra

##### 2.1 Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

addition = A + B

print("\nA + B:")
print(addition)



element_multiplication = A * B

print("\nElement-wise multiplication (A * B):")
print(element_multiplication)




matrix_multiplication = A @ B

print("\nMatrix multiplication (A @ B):")
print(matrix_multiplication)
```

##### 2.2 For a square matrix, find its transpose and determinant

```python

transpose_A = A.T

print("\nTranspose of A:")
print(transpose_A)


# ---------------------------------------------------------
# 6. Determinant
# ---------------------------------------------------------

det_A = np.linalg.det(A)

print("\nDeterminant of A:")
print(det_A)
```

##### 2.3 If invertible, find the inverse and verify A @ A^-1 using np.allclose()

```python
if not np.isclose(det_A, 0):
    inverse_A = np.linalg.inv(A)

    print("\nInverse of A:")
    print(inverse_A)

   
    result = A @ inverse_A

    print("\nA @ A^-1:")
    print(result)

    identity_matrix = np.eye(A.shape[0])

    print("\nIs A @ A^-1 equal to Identity Matrix?")
    print(np.allclose(result, identity_matrix))

else:
    print("\nMatrix A is not invertible because its determinant is zero.")
```

# Part B - Data Wrangling with Pandas: Titanic Dataset

## Task 4 - Load and Inspect Data

##### 4.1 Load train.csv. Implement head(), tail(), shape, columns, info(), and describe()

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("train.csv")

# Head
print("First 5 rows:")
display(df.head())

# Tail
print("Last 5 rows:")
display(df.tail())

# Shape
print("Shape:", df.shape)

# Columns
print("\nColumns:")
print(df.columns.tolist())

# Information
print("\nDataset Information:")
df.info()

# Descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe())
```

##### 4.2 Use loc and iloc to select rows/columns. Give one example of each and state the difference.


```python
# loc: label-based selection
print("Using loc:")
display(df.loc[0:4, ["Name", "Sex", "Age"]])

# iloc: integer-position-based selection
print("Using iloc:")
display(df.iloc[0:5, 0:5])
```

## Task 5 — Filtering and Querying

##### 5.1  Male passengers older than 50: how many?


```python
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

print("Male passengers older than 50:", len(male_over_50))
```

##### 5.2 Female first-class passengers: how many, and what percentage survived?

```python
female_first = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

female_first_count = len(female_first)

female_first_survival_rate = female_first["Survived"].mean() * 100

print("Female first-class passengers:", female_first_count)
print("Survival percentage:", female_first_survival_rate)
```

##### 5.3 Age 20–40, Fare above overall median, and survived

```python
fare_median = df["Fare"].median()

condition = df[
    (df["Age"] >= 20) &
    (df["Age"] <= 40) &
    (df["Fare"] > fare_median) &
    (df["Survived"] == 1)
]

print("Passengers:", len(condition))
```

##### 5.4 Travelling alone, age < 30, and did not survive

```python
condition = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

print("Passengers:", len(condition))
```

##### 5.5 Embarked S, Pclass 2 or 3, Fare above Southampton median

```python
southampton_median = df.loc[
    df["Embarked"] == "S", "Fare"
].median()

print("Southampton median fare:", southampton_median)

condition = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median)
]

print("Passengers:", len(condition))
```

## Task 6 - groupby() and Aggregation


##### 6.1 Survival rate by Sex.
##### 6.2 Survival rate by Pclass.
##### 6.3 Average Age and Fare by Pclass

```python
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("Survival rate by Sex:")
print(survival_by_sex)


survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100

print("Survival rate by Pclass:")
print(survival_by_class)


avg_age_fare = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("Average Age and Fare by Pclass:")
display(avg_age_fare)
```

##### 6.4 Passenger count and survival rate by Sex-Pclass

```python
sex_class = df.groupby(["Sex", "Pclass"]).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

sex_class["Survival_Rate"] *= 100

print("Sex-Pclass analysis:")
display(sex_class)
```

##### 6.5 Passenger count, average Fare, survival rate by Embarked

```python
embarked_stats = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

embarked_stats["Survival_Rate"] *= 100

print("Embarked statistics:")
display(embarked_stats)
```

## Task 7 - Missing Values and Fare Outliers

##### 7.1 Find the missing-value count and percentage for every column. Plot missing counts as a bar chart.


```python
import os

os.makedirs("figures", exist_ok=True)


missing_count = df.isnull().sum()

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

display(missing_table)

plt.figure(figsize=(10, 5))

missing_count[missing_count > 0].plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Column")
plt.ylabel("Missing Count")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("figures/missing_values.png")
plt.show()
```

##### 7.2 Fill missing Age values with the mean Age. Show missing counts before and after.


```python
print("Missing Age before:", df["Age"].isnull().sum())

age_mean = df["Age"].mean()

df["Age"] = df["Age"].fillna(age_mean)

print("Missing Age after:", df["Age"].isnull().sum())
```

##### 7.3 Try different imputation methods

```python
age_mean_df = df.copy()

age_mean_value = age_mean_df["Age"].mean()

age_mean_df["Age"] = age_mean_df["Age"].fillna(age_mean_value)


age_median_df = df.copy()

age_median_value = age_median_df["Age"].median()

age_median_df["Age"] = age_median_df["Age"].fillna(age_median_value)


age_mode_df = df.copy()

age_mode_value = age_mode_df["Age"].mode()[0]

age_mode_df["Age"] = age_mode_df["Age"].fillna(age_mode_value)




age_random_df = df.copy()

missing_mask = age_random_df["Age"].isna()

np.random.seed(42)

random_values = np.random.choice(
    age_random_df.loc[~missing_mask, "Age"],
    size=missing_mask.sum()
)

age_random_df.loc[missing_mask, "Age"] = random_values
```

##### 7.4 Fare Outliers using IQR

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", len(outliers))
```

## Task 8 - Features and Pivot Table


##### 8.1 Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.


```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

print(df[["SibSp", "Parch", "FamilySize"]].head())


df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

print(df[["FamilySize", "IsAlone"]].head())
```

##### 8.2 Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest groups.

```python
pivot_table = pd.pivot_table(
    df,
    index="Sex",
    columns="Pclass",
    values="Survived",
    aggfunc="mean"
)

print("Survival Rate Pivot Table:")
display(pivot_table)


highest_group = pivot_table.stack().idxmax()
highest_rate = pivot_table.stack().max()

lowest_group = pivot_table.stack().idxmin()
lowest_rate = pivot_table.stack().min()

print("Highest survival group:", highest_group)
print("Highest survival rate:", highest_rate * 100, "%")

print("Lowest survival group:", lowest_group)
print("Lowest survival rate:", lowest_rate * 100, "%")
```

##### 9.1 Correlation heatmap

```python
numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

correlation = df[numerical_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.savefig("figures/correlation_heatmap.png")
plt.show()
```

Find strongest relationships programmatically:

```python
corr_pairs = correlation.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) !=
    corr_pairs.index.get_level_values(1)
]

print("Strongest positive correlation:")
print(corr_pairs.sort_values(ascending=False).head(2))

print("\nStrongest negative correlation:")
print(corr_pairs.sort_values().head(2))
```
